# Mean Imputter  -SimpleImputer

```python
mean_imputer = SimpleImputer(missing_values=np.nan,strategy='mean')
df_1['CLMAGE'] = pd.DataFrame(mean_imputer.fit_transform(df_1[['CLMAGE']]))
```

```python
df_1.isna().sum()
```

```python
median_imputer = SimpleImputer(missing_values=np.nan,strategy='median')
df_1['CLMSEX'] = pd.DataFrame(median_imputer.fit_transform(df_1[['CLMSEX']]))
```

```python
df_1.isna().sum()
```

```python
mode_imputer = SimpleImputer(missing_values=np.nan,strategy='most_frequent')
df_1['CLMINSUR'] = pd.DataFrame(median_imputer.fit_transform(df_1[['CLMINSUR']]))
```

```python
df_1.isna().sum()
```

```python
mode_imputer = SimpleImputer(missing_values=np.nan,strategy='constant',fill_value=10)
df_1['SEATBELT'] = pd.DataFrame(median_imputer.fit_transform(df_1[['SEATBELT']]))
```

```python
df_1.isna().sum()
```

# fillna

```python
df_2.isna().sum()
```

```python
df_2['CLMSEX'].fillna(1,inplace=True)
df_2['CLMINSUR'].fillna(2,inplace=True)
df_2['SEATBELT'].fillna(3,inplace=True)
df_2['CLMAGE'].fillna(4,inplace=True)
```

```python
df_3['CLMSEX'].fillna(df_3['CLMSEX'].mean(),inplace=True)
df_3['CLMINSUR'].fillna(df_3['CLMINSUR'].median(),inplace=True)
df_3['SEATBELT'].fillna(df_3['SEATBELT'].mode()[0],inplace=True)
df_3['CLMAGE'].fillna(df_3['CLMAGE'].mean(),inplace=True)
```

```python
df_3.isna().sum()
```
