# LABEL ENCODING

```python
df_2.head()
```

```python
df_2.drop(columns=['Index'],inplace=True)
```

```python
df_2.columns
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lbc = LabelEncoder()
```

```python
df_2['Animals'] = lbc.fit_transform(df_2['Animals'])
```

```python
for col in df_2.columns:
    lbc = LabelEncoder()
    df_2[col] = lbc.fit_transform(df_2[col])
```

```python
df_2.head()
```

```python
df_2["Animals"] = lbc.inverse_transform(df_2['Animals'])
```

```python
df_2.head()
```

```python
mapping = list(
    zip(
        lbc.classes_,lbc.transform(lbc.classes_)
    )
)
mapping
```
