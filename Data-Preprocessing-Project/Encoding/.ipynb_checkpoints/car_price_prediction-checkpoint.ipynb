## Levy

### Identify the bad values

```python
(data['Levy'] == '-').sum()
```

```python
data['Levy'].isna().sum()
```

### Replace invalid placeholders with NaN

```python
data['Levy'] = data['Levy'].replace('-',np.nan)
```

```python
data['Levy'].isna().sum()
```

### Convert safely

```python
data['Levy'] = data['Levy'].astype('float64')
```

```python
data['Levy'].info()
```

### Filling missing value with mean()

```python
data['Levy'].fillna(data['Levy'].mean(),inplace=True)
```

```python
data['Levy'].isna().sum()
```

```python
data.select_dtypes(include='object').nunique()
```

## Gear box type , Drive wheels

```python
data['Drive wheels'].unique()
```

```python
dummies = pd.get_dummies(data['Drive wheels'],prefix='Drive', drop_first=False)
```
