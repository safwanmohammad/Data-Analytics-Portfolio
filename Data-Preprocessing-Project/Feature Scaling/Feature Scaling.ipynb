## StandardScaler

```python
# usally around -3 to 3 range
```

```python
df_3.head()
```

```python
def standardscal(i):
    x = ( i - i.mean()) / i.std()
    return x

standardscal(df_3).head()
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
x = ss.fit_transform(df_3)
df_3 = pd.DataFrame(x,columns=data.columns)
df_3.head()
```

# Robust_Scaler

```python
df_4.head()
```

```python
def robust_scaler(x):
    q1 = x.quantile(0.25)
    q3 = x.quantile(0.75)
    iqr = q3 - q1
    new_rs = (x - x.median())  / iqr
    return new_rs

robust_scaler(df_4).head()
```

```python
from sklearn.preprocessing import RobustScaler
```

```python
scaler = RobustScaler()
df_4 = scaler.fit_transform(df_4)
df_4 = pd.DataFrame(df_4,columns=data.columns)
df_4.head()
```
