# Axis_Labels

```python
plt.plot(years,yild_apples,color='g',marker='*')
plt.xlabel('YEARS')
plt.ylabel('YILD_APPLES')
plt.show()
```

# Adding title and Marker

```python
plt.title('Sales Distribution over Year',color='r')
plt.plot(years,yild_apples,marker='*')
plt.show()
```

```python
plt.title('Sales Distribution over Year',color='r')
plt.xlabel("YEARS",color='g')
plt.ylabel("yild_apples",color='g')
plt.plot(years,yild_apples,marker='o')
plt.show()
```

# customizing size total grap

```python
plt.figure(figsize=(10,3),facecolor='g',edgecolor='r')
plt.title('Sales Distribution over Year',color='r')
plt.xlabel("YEARS",color='g')
plt.ylabel("yild_apples",color='g')
plt.plot(years,yild_apples,marker='o')
plt.show()
```

# x and y labels arrangements

```python
plt.figure(figsize=(10,3))
plt.title('Sales Distribution over Year',color='r')
plt.xlabel("YEARS",color='g')
plt.ylabel("yild_apples",color='g')
plt.plot(years,yild_apples,marker='o')
plt.xticks(rotation=45)
plt.yticks(rotation = 45)
plt.show()
```

# grid

```python
plt.figure(figsize=(10,3))
plt.title('Sales Distribution over Year',color='r')
plt.xlabel("YEARS",color='g')
plt.ylabel("yild_apples",color='g')
plt.plot(years,yild_apples,marker='o')
plt.xticks(rotation=45)
plt.yticks(rotation = 45)
plt.grid()
plt.show()
```

#  BAR CHART

```python
x = ['Shirts', 'Pants','Shorts','Shoes']
y1 = [1000, 1200, 800, 1800]
y2 = [500,600,400,900]
len(y2)
```

```python
plt.figure(figsize=(10,4))
plt.xlabel('Garments')
plt.ylabel('Sales')
plt.title('Sales Reporte')
plt.xticks(rotation = 45)
plt.yticks(rotation = 45)
plt.bar(x,y1,color='g')
plt.bar(x,y2,color='r' ,bottom=y2,)
plt.show()
```

# plt.bar(X,Y) WITH plt.barh(X,Y)

```python
plt.figure(figsize=(10,4))
plt.xlabel('Garments')
plt.ylabel('Sales')
plt.title('Sales Reporte')
plt.xticks(rotation = 45)
plt.yticks(rotation = 45)
plt.grid(True)
plt.barh(x,y1,color='r')
plt.show()
```

# STACKED BAR

```python
x = ['Shirts', 'Pants','Shorts','Shoes']
y1 = [1000, 1200, 800, 1800]
y2 = [500,600,400,900]
```

```python
plt.title('2_value_Stack')
plt.bar(x,y1,color='r')
plt.bar(x,y2)
plt.xlabel("Dress")
plt.ylabel("Sales_revenue")
plt.legend(['y1','y2'],loc='best')
plt.show()
```

```python
import pandas as pd
data = {
    'Qurter' : ['q1','q2','q3','q4'],
    'product A':[10,20,30,40],
    'product B':[20,25,35,45],
    'product C':[30,30,40,50]
}
df = pd.DataFrame(data)
df
```

```python
# now its look like table
df1 = df.set_index("Qurter")
df1 
```

```python
# plot stacked bar chart
df1.plot(kind='area',stacked='True',figsize=(10,4),colormap='plasma')
plt.title("Sales Report")
plt.xlabel('Qurters')
plt.ylabel("Sales")
plt.legend(title='Products')
plt.show()
```

```python
# plot stacked bar chart
df1.plot(kind='bar',stacked='True',figsize=(10,4),colormap='viridis')
plt.title("Sales Report")
plt.xlabel('Qurters')
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.legend(title='Products')
plt.show()
```

# PIE CHART

```python
# Data to plot
labels = ['Marketing','Reserch & Development','Sales','Hr']
# The size for each slice(number of empployees in each department)
sizes = [20,30,25,35]
# custom colors
colors = ['gold','silver','yellow','skyblue']
```

```python
plt.pie(
    sizes,
    explode = (0,0,0.1,0),
    labels = labels,
    colors = colors,
    autopct = '%1.1f%%',
    shadow = True,
    startangle = 10
)
plt.title('Number Of Employees Across Department')
```

```python
data = {
'season': ['Summer', 'Summer', 'Winter', 'Winter'],
'product': ['Heater', 'AC', 'Heater', 'AC'],
'count': [16, 40, 32, 13]
}
df = pd.DataFrame(data)
df
```

```python
season_count = df.groupby('season')['count'].mean()
season_count
```

```python
season_count.plot(kind='pie',colors=['gold','silver'],explode=(0,0.1),shadow=True,autopct='%1.1f%%')
```

# BOX plot

```python
data = {'windspeed' : [10, 15, 8, 20, 12, 0, 7, 14, 21, 18, 9, 6, 3, 11, 5, 13, 17, 19, 22, 16, 8, 10, 7, 5, 13,]}
df = pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(10,5))
plt.boxplot(df['windspeed'],patch_artist=True)
plt.title('Box plot of wind speed')
plt.xlabel('wind speed')
plt.ylabel('value')
plt.show()
```

```python
df['windspeed'].mean()
```
