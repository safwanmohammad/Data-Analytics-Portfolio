What is Covariance?
Covariance measures how two random variables change together. It is calculated by averaging the product of their deviations from their means. A positive value means they move in the same direction, while a negative value means they move in opposite directions.

It can take any value between - infinity to +infinity, where the negative value represents the negative relationship whereas a positive value represents the positive relationship.
It is used for the linear relationship between variables.
It gives the direction of relationship between variables.

What is Correlation?
Correlation is a standardized measure of the strength and direction of the linear relationship between two variables. It is derived from covariance and ranges between -1 and 1. Unlike covariance, which only indicates the direction of the relationship, correlation provides a standardized measure.

Positive Correlation (close to +1): As one variable increases, the other variable also tends to increase.
Negative Correlation (close to -1): As one variable increases, the other variable tends to decrease.
Zero Correlation: There is no linear relationship between the variables.

# covariation 

```python
import pandas as pd
import numpy as np
import seaborn as sns
import warnings
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')
```

```python
df = pd.DataFrame(
    {
    'X': np.arange(1,6),
    'Y': np.arange(2,12,2)
    }
)
df
```

```python
np_1 = np.cov(df.X,df.Y)
np_1
```

```python
np_1 = np.cov(df.X,df.Y)[0,1]
np_1
np_1 = np.cov(df.X,df.Y)
np_1
```

```python
sns.heatmap(np_1,
            annot=True,      # Show correlation values on the map
            cmap='viridis', # Color map (e.g., coolwarm, RdYlBu, viridis)
            fmt=".2f",       # Format annotations to 2 decimal places
            linewidths=.5) 
plt.title('Correlation Matrix Heatmap')
plt.show()
```

# correlation

```python
np_2 = df.corr()
np_2
```

```python
sns.heatmap(np_2,
            annot=True,      # Show correlation values on the map
            cmap='RdYlBu', # Color map (e.g., coolwarm, RdYlBu, viridis)
            fmt=".2f",       # Format annotations to 2 decimal places
            linewidths=.5) 
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
# Load the built-in 'tips' dataset
tips = sns.load_dataset('tips')
tips
```

```python
# Compute Correlatiom Matrix
corr = tips[["total_bill","tip","size"]].corr()
corr
```

```python
sns.heatmap(corr, 
            annot=True,      # Show correlation values on the map
            cmap='coolwarm', # Color map (e.g., coolwarm, RdYlBu, viridis)
            fmt=".2f",       # Format annotations to 2 decimal places
            linewidths=.5) 
plt.title('Correlation Matrix Heatmap')
plt.show()
```
