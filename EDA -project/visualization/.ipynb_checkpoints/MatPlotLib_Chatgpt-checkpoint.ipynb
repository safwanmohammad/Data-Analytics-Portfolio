# MATPLOTLIB

# 1️⃣ CORE STRUCTURE (UNDERSTAND THIS OR STOP)

Every matplotlib chart is built from 3 things:

Data → what you’re plotting

Axes → where it’s plotted

Visual encoding → how values are shown (position, length, color)

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Basic Line Chart")
plt.show()
```

# 2️⃣ THE 6 CHARTS YOU MUST MASTER (NO EXCUSES)
## 1. LINE CHART – trends over time
Use when:

Time series

Ordered data

❌ Never use when:

Categories like names, products, cities

```python
plt.plot([1,2,3,4], [10,20,15,25])
plt.show()
```

## 2.BAR CHART – comparing categories

Question it answers:

“Which category is bigger or smaller?”
If you use a line chart here → you don’t understand data visualization.

```python
names = ["A", "B", "C"]
values = [30, 20, 40]

plt.bar(names, values)
plt.show()
```

## 3. HISTOGRAM – distribution of values

Question it answers:

“How are values spread?”
Key concept:

Histogram ≠ bar chart

Histogram bins numeric ranges, not categories

Interview trap:
If you say “histogram shows frequency of categories” → instant rejection.

```python
data = [1,2,2,3,3,3,4,5,6,7,8]

plt.hist(data, bins=5)
plt.show()
```

## 4. BOX PLOT – spread + outliers (VERY IMPORTANT)

Question it answers:

“What’s the distribution, median, and outliers?”

Why professionals love box plots:

You compare multiple distributions easily

Outliers are obvious

If you don’t understand box plots → you’re weak in statistics.

```python
data = [10,12,14,15,18,19,100]

plt.boxplot(data)
plt.show()
```

## 5. SCATTER PLOT – relationships

Question it answers:

“Are two variables related?”

Important:

Scatter ≠ trend

Scatter shows patterns, not conclusions

If you say “this proves correlation” → wrong. It only suggests it.

```python
x = [1,2,3,4,5]
y = [2,4,5,4,6]

plt.scatter(x, y)
plt.show()
```

## 6. HEATMAP – intensity / correlation matrices

Question it answers:

“Where are values high or low in a matrix?”

Used for:

Correlation matrices

Model diagnostics

Large numeric tables

```python
import numpy as np

data = np.random.rand(5,5)

plt.imshow(data)
plt.colorbar()
plt.show()
```

# 3️⃣ HOW TO CHOOSE THE RIGHT CHART (THIS IS GOLD)



Ask ONE question:

Is the x-axis ordered?

YES → Line chart

NO → Bar chart

Are you showing distribution?

YES → Histogram / Box plot

Two numeric variables?

YES → Scatter

Matrix or correlation?

YES → Heatmap

If you hesitate here → you don’t understand the problem yet.
