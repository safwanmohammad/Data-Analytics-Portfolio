# Bar Chart

When comparing values between and/or among groups, bar plots are traditionally the visualization of choice.

```python
x = ['Shirts','Pants','Shorts','shoes','tittes']
y1 = np.random.randint(100,500,5)
```

```python
plt.bar(x,y1)
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.bar(x,y1,color='red')
plt.title("Sale Distribution of y1 and y2")
plt.xlabel("X")
plt.ylabel("Y1 and Y2")
plt.legend(['y1','y2'],loc = 'upper right')
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.grid()
plt.show()
```

If we wish to convert the plot into a horizontal bar plot, we can do so by replacing.
# plt.bar(x, y) with plt.barh(x, y)

```python
plt.figure(figsize=(10,4))
plt.barh(x,y1,color='red')
plt.title("Sale Distribution of y1 and y2")
plt.xlabel("X")
plt.ylabel("Y1 and Y2")
plt.legend(['y1','y2'],loc = 'upper right')
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.grid()
plt.show()
```

# Pie Chart
In a pie chart, each slice stands for a different category, and the size of the slice shows how much of the
Consider a company's budget where the spending is divided among different departments
Marketing
Research & Development
Sales and Operations
HR
A pie chart could effectively show how much of the total budget each department consumes, allowing st spending in each area.

```python
# Data to plot
labels = ['Marketing', 'Research & Development', 'Sales', 'HR']
# The sizes for each slice(no of employees in each department)
sizes = [20, 30, 25, 25]
#Optional: custom colors
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
```

```python
plt.pie(sizes,explode=(0,0.1,0,0),labels=labels,colors=colors,autopct='%1.1f%%',shadow=True,startangle=10)
plt.show()
```

```python
data = {
'season': ['Summer', 'Summer', 'Winter', 'Winter'],
'product': ['Heater', 'AC', 'Heater', 'AC'],
'count': [16, 40, 32, 13]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
season_count = df.groupby('season')['count'].mean()
season_count
```

```python
season_count.plot(kind='pie',colors=['red','blue'],explode=(0,0.1),shadow=True,autopct='%1.f1%%')
plt.show()
```

# histogram
A histogram is a statistical graph that represents the distribution of a continuous dataset through plotted bars, each representing a particular category or class interval. The bar height reflects the frequency or count of data points within each group.

```python
x = [1, 2, 2, 2, 2, 2, 2, 3, 3, 3, 4, 5, 5, 6, 6, 6, 7, 7, 7, 7, 7, 7]
#here y is the occurance of the number or count of number
plt.hist(x,bins=7)
plt.show()
```

```python
# basic histogram
import matplotlib.pyplot as plt
# Data (Example: Randomly generated data)
data= [1, 2, 1, 3, 3, 3, 4, 5, 5, 6, 7, 8, 8, 9, 10]
# Create the histogram
plt.hist(data, bins=9, color='skyblue', edgecolor='black')
# Add title and Labels
plt.title("Basic Histogram")
plt.xlabel("Value Ranges")
plt.ylabel("Frequency")
# Show the plot
plt.show()
```

```python
humidity_data = np.random.randint(20,100,100)
```

# Box-and-whisker
Another common statistical plot for exploring a numerical feature is a box-and-whisker plot, also referred to as a boxplot.
Box-and-whisker plots display the distribution of an array based on the minimum, first quartile, median, third quartile, and maximum, but they are primarily used to indicate the skew of a distribution and to identify outliers.

```python
data = {
    'windspeed' : [10, 15, 8, 20, 12, 0, 7, 14, 21, 18, 9, 6, 3, 11, 5, 13, 17, 19, 22, 16, 8, 10, 7, 5, 13,]
}
df = pd.DataFrame(data)
df
```

```python
df.windspeed.mean()
```

```python
plt.figure(figsize=(8,6))
plt.boxplot(df['windspeed'],patch_artist=True)
plt.title("Box plot of Windspeed")
plt.xlabel("Wind Speed")
plt.ylabel("Value")
plt.grid()
plt.show()
```

```python
data = {
    'windspeed' : [10, 15, 8, 20, 12, 0, 7, 14, 21, 18, 9, 6, -13, 11, 5, 13, 17, 19, 22, 16, 8, 10, 7, 5, 35]
}
df = pd.DataFrame(data)
df
```

```python
df.windspeed.mean()
```

```python
plt.figure(figsize=(8,6))
plt.boxplot(df['windspeed'],patch_artist=True)
plt.title("Box plot of Windspeed")
plt.xlabel("Wind Speed")
plt.ylabel("Value")
plt.grid()
plt.show()
```

# subplot
subplots method provides a way to plot multiple plots on a single figure.
1. plt.subplot(row number, total number of columns, position of the graph)

```python
x1=[1,5,2,4]
y1=[4,1,5,3]
plt.subplot(1,4,1)
plt.plot(x1,y1)
plt.title("[1,4,1]")

x2=[1,5,2,4]
y2=[4,1,5,3]
plt.subplot(1,4,2)
plt.bar(x2,y2)
plt.title("[1,4,2]")

x3=[1,5,2,4]
y3=[4,1,5,3]
plt.subplot(1,4,3)
plt.barh(x1,y1)
plt.title("[1,4,3]")

x4=[1,5,2,4]
y4=[4,1,5,3]
plt.subplot(1,4,4)
plt.scatter(x1,y1)
plt.title("[1,4,4]")

plt.subplots_adjust(wspace=0.5,hspace=0)
plt.show()
```

```python
x1=[1,5,2,4]
y1=[4,1,5,3]
plt.subplot(2,2,1)
plt.plot(x1,y1)
plt.title("[1,4,1]")

x2=[1,5,2,4]
y2=[4,1,5,3]
plt.subplot(2,2,2)
plt.bar(x2,y2)
plt.title("[1,4,2]")

x3=[1,5,2,4]
y3=[4,1,5,3]
plt.subplot(2,2,3)
plt.barh(x1,y1)
plt.title("[1,4,3]")

x4=[1,5,2,4]
y4=[4,1,5,3]
plt.subplot(2,2,4)
plt.scatter(x1,y1)
plt.title("[1,4,4]")

plt.subplots_adjust(wspace=0.5,hspace=1)
plt.show()
```
