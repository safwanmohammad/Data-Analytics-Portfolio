#  BAR CHART

```python
x = ['Shirts', 'Pants','Shorts','Shoes']
y1 = [1000, 1200, 800, 1800]
y2 = [500,600,400,900]
len(y2)
```

```python
plt.figure(figsize=(10,4))
plt.xlabel('Garments')
plt.ylabel('Sales')
plt.title('Sales Reporte')
plt.xticks(rotation = 45)
plt.yticks(rotation = 45)
plt.bar(x,y1,color='g')
plt.bar(x,y2,color='r' ,bottom=y2,)
plt.show()
```

# plt.bar(X,Y) WITH plt.barh(X,Y)

```python
import pandas as pd
data = {
    'Qurter' : ['q1','q2','q3','q4'],
    'product A':[10,20,30,40],
    'product B':[20,25,35,45],
    'product C':[30,30,40,50]
}
df = pd.DataFrame(data)
df
```

```python
# now its look like table
df1 = df.set_index("Qurter")
df1 
```

```python
# plot stacked bar chart
df1.plot(kind='area',stacked='True',figsize=(10,4),colormap='plasma')
plt.title("Sales Report")
plt.xlabel('Qurters')
plt.ylabel("Sales")
plt.legend(title='Products')
plt.show()
```

```python
# plot stacked bar chart
df1.plot(kind='bar',stacked='True',figsize=(10,4),colormap='viridis')
plt.title("Sales Report")
plt.xlabel('Qurters')
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.legend(title='Products')
plt.show()
```

# PIE CHART

```python
# Data to plot
labels = ['Marketing','Reserch & Development','Sales','Hr']
# The size for each slice(number of empployees in each department)
sizes = [20,30,25,35]
# custom colors
colors = ['gold','silver','yellow','skyblue']
```

```python
plt.pie(
    sizes,
    explode = (0,0,0.1,0),
    labels = labels,
    colors = colors,
    autopct = '%1.1f%%',
    shadow = True,
    startangle = 10
)
plt.title('Number Of Employees Across Department')
```

```python
data = {
'season': ['Summer', 'Summer', 'Winter', 'Winter'],
'product': ['Heater', 'AC', 'Heater', 'AC'],
'count': [16, 40, 32, 13]
}
df = pd.DataFrame(data)
df
```

# JOIN PLOT

```python
# To compain histogram and scatter its called join plot 
# also using seaborn module
```

```python
import seaborn as sns
```

```python
data  = pd.DataFrame(
    {
        'x' : np.arange(1,11),
        'y' : np.arange(3,33,3)
    }
)
data
```

```python
sns.jointplot(data=data,x='x',y='y',kind='scatter')
plt.grid()
plt.show()
```

```python
data  = pd.DataFrame(
    {
        'x' : np.arange(1,11),
        'y' : [3,1,42,1,42,31,54,23,5,12]
    }
)
data
sns.jointplot(data=data,x='x',y='y',kind='scatter')
plt.grid()
plt.show()
```
