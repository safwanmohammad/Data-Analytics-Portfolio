# Merge

```python
d1 = pd.DataFrame(
    {
        'empid':[101,102,103,104,105],
        'empname':['A','B','C','D','E'],
        'designation':['Doct','eng','adv','tea','dev']
    }
)
d1
```

```python
d2 = pd.DataFrame(
    {
        'empid':[101,102,103,106,107],
        'salary':[1000,2000,3000,4000,5000],
        'lacaton':['CL','IN','GM','UAE','UK']
    }
)
d2
```

## INNER JOIN

```python
inner = pd.merge(d1,d2,how='inner',on='empid')
inner
```

## OUTER JOIN

```python
outer = pd.merge(d1,d2,how='outer',on='empid')
outer
```

## RIGHT JOIN

```python
right_join = pd.merge(d1,d2,how='right',on='empid')
right_join
```

## LEFT JOIN

```python
left_join = pd.merge(d1,d2,how='left',on='empid')
left_join
```

```python
pd.merge(d1,d2,how='right',on='empid')
```
