```python
# 1.Give column name as Id, Movie, Year, Rating, Duration.

df = pd.read_csv(r'C:\Users\babus\Downloads\movies_cleaned (1).csv',
                 names=[ 'Id','Movie', 'Year','Rating', 'Duration']
                )
df
```

```python
# 2.Display the number of rows and number of columns.
df.info()
```

```python
# 3.Display the five observations that have highest duration.
x = df.sort_values(by='Duration',ascending = False)
x.head()
```

```python
# 4.Remove if any duplicate exists and change the index accordingly.
df.duplicated().sum()
df = df.drop_duplicates(ignore_index=True)
df
```

```python
df.sort_values(by='Rating').head()
```

```python
# 5.Rename the column name ‘Movie’ as Film.
x = df.rename(columns={'Movie':'Film'})
x
# y = df.drop(columns=[{'':0}])
# # y
```

```python
# 6.Display all the observation starting from index 100 upto 200.
df.iloc[100:201]
```

```python
# 7.Display the name and releasing year of the film with maximum duration.
```

```python
df.sort_values(by='Duration',ascending = False).loc[
df['Duration'] == df['Duration'].max() ,['Movie','Year']   
]
```

```python
# 8.Display the count of films released in each year.
df.loc[
    df['Year'] == df['Year']
]
# x = df['Year'].unique()
# x
x = df['Year'].value_counts().sort_index()
x
```

```python
y = df[['Year']].value_counts()
y
```

```python
# 9.Find the film released between 1986–1996 having highest duration.
df.loc[
    (df['Year'] >= 1986) &
    (df['Year'] <= 1996)
].sort_values(by='Duration',ascending = False).head(1)
```

```python
# 10 Find the films having duration greater than 7000 and rating above 3.5.
df.loc[
    (df['Duration'] > 7000) &
    (df['Rating'] >= 3.5 )
]
```
