# 3️⃣ Element Accessing & Slicing

```python
# From a 1D array [5,10,15,20,25]:
x = np.arange(5,26,5)
x
# Access the last element
x[-1]
# Slice [10,15,20]
x[1:4]
```

```python
# From a 2D array (3×3):
x = np.arange(1,10).reshape(3,3)
x
# Extract the second row
x[1:2]
# Extract the first column
x[:,0:1]
# Extract the center element
x[1:2,1:2]
```

```python
# From a 3D array:
x = np.arange(1,13).reshape(2,2,3)
x
# Access a single element
x[1:2,0:1,1:2]
# Access an entire 2D slice
x[0,:,:]
```

# 6️⃣ Merge & Concat

```python
d1 = pd.DataFrame(
    {
        'name':['A','B'],'roll_no':[1,2]
    }
)
d2 = pd.DataFrame(
    {
        'name':['C','D'],'roll_no':[99,98]
    }
)
# Concatenate two DataFrames:
df = pd.concat([d1,d2])
df
```

```python
# Row-wise
df = pd.concat([d1,d2],axis=0)
df
```

```python
# Column-wise
df = pd.concat([d1,d2],axis=1)
df
```

```python
# Reset index after concatenation.
df = pd.concat([d1,d2],ignore_index=True)
df
```

```python
# Perform:
d1 = pd.DataFrame(
    {
        'empid':[101,102,103,104,105],
        'empname':['A','B','C','D','E'],
        'designation':['Doct','eng','adv','tea','dev']
    }
)
d2 = pd.DataFrame(
    {
        'empid':[101,102,103,106,107],
        'salary':[1000,2000,3000,4000,5000],
        'lacaton':['CL','IN','GM','UAE','UK']
    }
)
d1,d2
```

```python
# inner merge
inner = pd.merge(d1,d2,how='inner',on='empid')
inner
```

```python
# left merge
left = pd.merge(d1,d2,how='outer',on='empid')
left
```

```python
# right merge
right = pd.merge(d1,d2,how='right',on='empid')
right
```

```python
# outer merge
out = pd.merge(d1,d2,how='outer',on='empid')
out
```

```python
# Identify which rows appear only in an outer merge.
# all rows with nan value
```
