# Crim outliers

```python
import numpy as np
```

```python
q1 = df['crim'].quantile(0.25)
q3 = df['crim'].quantile(0.75)
IQR = q3 - q1 

lower_limit = q1 - (IQR * 1.5)
upper_limit = q3 + (IQR * 1.5)
```

```python
q1
```

```python
q3
```

```python
IQR
```

```python
lower_limit
```

```python
upper_limit
```

```python
outliers  = np.where(df['crim'] > upper_limit,True,np.where(df['crim'] < lower_limit,True,False))
outliers
outliers_1 = (  df['crim'] < lower_limit  ) | ( df['crim'] > upper_limit )
outliers_1
```

```python
trimmed_crim = df['crim'].loc[~outliers]
trimmed_crim
trimmed_crim_1 = df['crim'].loc[~outliers_1]
trimmed_crim_1
```

```python
df['crim']
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,3,1)
plt.boxplot(df['crim'],patch_artist=True)
plt.title("_crim_")
plt.subplot(1,3,2)
plt.boxplot(trimmed_crim,patch_artist=True)
plt.title("__trimmed_crim__")
plt.subplot(1,3,3)
plt.boxplot(trimmed_crim_1,patch_artist=True)
plt.title("__trimmed_crim_1__")
plt.show()
```

```python
# This forces everything into the range, leaving 0 visual outliers
df['crim'] = df['crim'].clip(lower=lower_limit, upper=upper_limit)

plt.boxplot(df['crim'],patch_artist = True)
plt.show()

```

```python
q1 = df['zn'].quantile(0.25)
q1
```

```python
q3 = df['zn'].quantile(0.75)
q3
```

```python
IQR = q3 - q1
IQR
```

```python
LOWER_LIMIT = q1 - (IQR * 1.5)
LOWER_LIMIT
```

```python
UPPER_LIMIT = q3 + (IQR * 1.5)
UPPER_LIMIT
```

```python
outliers = ( df['zn'] > UPPER_LIMIT) & (df['zn'] < LOWER_LIMIT )
outliers
outliers_1 = ( df['zn'] > UPPER_LIMIT) | (df['zn'] < LOWER_LIMIT )
outliers_1
```

```python
trimmed_zn = df['zn'].loc[~outliers]
trimmed_zn
trimmed_zn_1 = df['zn'].loc[~outliers_1]
trimmed_zn_1
```

```python
plt.boxplot(trimmed_zn,patch_artist=True)
plt.title("&")
plt.show()
```
